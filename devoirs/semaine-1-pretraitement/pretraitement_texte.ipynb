```python
import spacy
from nltk.stem import SnowballStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# 1. Préparation des outils
try:
    nlp = spacy.blank("xx")
except:
    nlp = spacy.blank("en")

stemmer = SnowballStemmer("french")

# Texte brut en français
text_brut = "Les chercheurs examinent l'intelligence mince. Ils analysent de nombreuses données."

# --- STEP 1 : Tokenisation & Stopwords ---
doc = nlp(text_brut.lower())

french_stopwords = ["le", "la", "les", "l", "un", "une", "des", "de", "du", "et", "ou", "mais", "donc", "car", "ni", "or", "il", "elle", "ils", "elles", "je", "tu", "nous", "vous", "ce", "cet", "cette", "ces", "lui", "leur", "leurs", "y", "en", "a", "au", "aux", "avec", "sans", "pour", "par", "dans", "sur", "sous", "chez", "entre", "parmi", "contre"]

tokens_clean = [t for t in doc if not t.is_punct and t.text not in french_stopwords]

print(f"1. Après Stopwords : {[t.text for t in tokens_clean]}")

# --- STEP 2 : Stemming (Racines) ---
stems = [stemmer.stem(t.text) for t in tokens_clean]
print(f"2. Stemming (Racines) : {stems}")

# --- STEP 3 : Lemmatisation ---
lemmas = [t.text for t in tokens_clean]
print(f"3. Lemmatisation : {lemmas}")

# --- STEP 4 : TF-IDF ---
corpus = [" ".join(lemmas), "intelligence mince technologie", "analyse de données"]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(corpus)

print("\n4. Poids TF-IDF (Importance des mots) :")
df = pd.DataFrame(tfidf_matrix.toarray(), columns=vectorizer.get_feature_names_out())
print(df.iloc[0].sort_values(ascending=False))
```
